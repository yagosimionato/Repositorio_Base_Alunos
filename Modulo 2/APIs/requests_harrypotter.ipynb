# harry potter API

```python
import requests

url ='https://hp-api.onrender.com/api/characters'
res = requests.get(url) # armazenamos o retorno
print(res)
print(type(res))



# Converter a resposta para json, mostrar na tela o resultado da conversão e o tipo da variavel contendo essa conversão
dados = res.json()
print(dados)
print(type(dados))



# Quantos personagens existem?
len(dados)



# Cada item da lista é um dicionario contendo as informações dos personagens,
# mosre todas as chaves do dicionario que está na 
# posição 0 da lista

for i in dados[0].keys():
    print(i)



for i in dados:
    if i ['house'] == 'Gryffindor':
        print(i['name'])



from random import randint
url ='https://hp-api.onrender.com/api/characters'
res = requests.get(url).json() # armazenamos o retorno


personagem = res[randint(0,len(res))]
print('Pesonagem aleatório de Harry Potter: ')
print('nome', personagem['name'])
print('Casa', personagem['house'])

```
